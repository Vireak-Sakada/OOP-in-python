```python
#import tool for abstraction
from abc import ABC, abstractmethod
#Abstract class
class Animal(ABC):

  def sound(self):
    pass
class Dog(Animal):
  #must impliment sound
  def sound(self):
    print("wooff")
#child class
class Cat(Animal):
  def sound(self):
    print("Meow")
#crate the object
dog = Dog()
cat = Cat()
dog.sound()
cat.sound()
```

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def go(self):
      pass
    @abstractmethod
    def stop(self):
      pass
class Car(Vehicle):
    def go(self):
      print("You drive the car")
    def stop(self):
      print("you stop the car")
# carte the object
# car = Car()
car.go()
car.stop()
class Motorcycle(Vehicle):
  def go(self):
    print("you ride the motorcycle")
  def stop(self):
    print("you stop the motorcycle")
#crate the object
#motor = Motorcycle()
#motor.go()
#motor.stop()
class boat(Vehicle):
  def go(self):
    print("you ride the boat ")
  def stop(self):
    print("you stop the boat ")
Boat = boat()
Boat.go()
Boat.stop()

```
